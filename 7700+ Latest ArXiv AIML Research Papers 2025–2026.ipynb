## AI/ML Research Papers Exploratory Data Analysis (EDA)

```
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

## Introduction

This project performs Exploratory Data Analysis (EDA) on an AI/ML research papers dataset.  
The dataset contains information about paper titles, authors, categories, publication years, abstracts, DOI references, and collaborations.  

The purpose of this analysis is to identify trends, patterns, and relationships in AI and Machine Learning research publications using statistical analysis and data visualization techniques.

```
df = pd.read_csv("/content/arxiv_ai_ml_papers.csv")
```

## Dataset Overview

The dataset contains AI and Machine Learning research papers collected from arXiv.  
It includes information such as paper titles, authors, categories, publication dates, abstracts, DOI references, and collaboration details.  

This dataset helps analyze research trends, publication growth, author contributions, and relationships between different research features using Exploratory Data Analysis (EDA).

## Research Paper Analysis

## Which primary categories are most common?

```
df["primary_category"].value_counts().head(10)
```

##  Top AI/ML Categories Bar Chart

```
df["primary_category"].value_counts().head(10).plot(kind="bar")

plt.xlabel("Primary Category")
plt.ylabel("Count")
plt.title("Top AI/ML Categories")

plt.show()
```

## Author Analysis

## Which authors published the highest number of papers?

```
df["first_author"].value_counts().head(10)
```

## Top Authors Bar Chart


```
df["first_author"].value_counts().head(10).plot(kind="bar")

plt.xlabel("First Author")
plt.ylabel("Number of Papers")
plt.title("Top Authors")

plt.show()
```

## Yaer Analysis

##  How many papers were published each year?

```
df["year"].value_counts().sort_index()
```

## Research Papers Published Per Year Line Chart

```
df["year"].value_counts().sort_index().plot(kind="line")

plt.xlabel("Year")
plt.ylabel("Number of Papers")
plt.title("Research Papers Published Per Year")

plt.show()
```

## Abstract Analysis

##  What is the average abstract length?

```
df["abstract_length"].mean()
```

## Abstract Length Distribution Histogram

```
df["abstract_length"].hist(bins=20)

plt.xlabel("Abstract Length")
plt.ylabel("Frequency")
plt.title("Abstract Length Distribution")

plt.show()
```

## Word Count Analysis


##  What is the average word count of papers?

```
df["word_count"].mean()
```

## Word Count Distribution Histogram


```
df["word_count"].hist(bins=20)

plt.xlabel("Word Count")
plt.ylabel("Frequency")
plt.title("Word Count Distribution")

plt.show()
```

## DOI Analysis

##  How many papers have DOI references?

```
df["has_doi"].value_counts()
```

## DOI Availability Pie Chart

```
df["has_doi"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.ylabel("")
plt.title("DOI Availability")

plt.show()
```

## Collaboration Analysis

##  How many papers involve large collaborations?

```
df["is_large_collaboration"].value_counts()
```

## Large Collaboration Papers Bar Chart

```
df["is_large_collaboration"].value_counts().plot(kind="bar")

plt.xlabel("Large Collaboration")
plt.ylabel("Count")
plt.title("Large Collaboration Papers")

plt.show()
```

## Relatonship Analysis

##  Is there any relationship between number of authors and word count?

## Authors vs Word Count Scatter Plot

```
plt.scatter(df["num_authors"], df["word_count"])

plt.xlabel("Number of Authors")
plt.ylabel("Word Count")
plt.title("Authors vs Word Count")

plt.show()
```

## Abstract Length vs Word Count Scatter Plot

```
plt.scatter(df["abstract_length"], df["word_count"])

plt.xlabel("Abstract Length")
plt.ylabel("Word Count")
plt.title("Abstract Length vs Word Count")

plt.show()
```

## Heatmap

## What patterns can be observed from the correlation heatmap?

## Correlation Heatmap

```
import seaborn as sns

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
